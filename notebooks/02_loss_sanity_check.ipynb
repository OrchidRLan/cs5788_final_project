# Loss Sanity Check
Verify L_SDF, L_normal, L_eikonal compute correctly on dummy data.

```python
import sys; sys.path.insert(0,'..')
import torch
from triposg_finetune.loss import TotalLoss

B, N = 2, 512
points   = torch.rand(B, N, 3, requires_grad=True)
gt_sdf   = torch.rand(B, N) * 2 - 1
gt_norm  = torch.randn(B, N, 3)
gt_norm  = gt_norm / gt_norm.norm(dim=-1, keepdim=True)
surf_mask= torch.rand(B, N) > 0.5
pred_sdf = (points ** 2).sum(dim=-1).sqrt() - 0.5  # sphere SDF

criterion = TotalLoss(lambda_normal=0.1, lambda_eikonal=0.05)
loss, metrics = criterion(pred_sdf, points, gt_sdf, gt_norm, surf_mask)
print('Loss computed successfully:')
for k, v in metrics.items(): print(f'  {k}: {v:.6f}')
```
