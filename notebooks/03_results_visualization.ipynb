# Results Visualization
Plot loss curves and qualitative comparisons.

```python
import json
import matplotlib.pyplot as plt

with open('../logs/train.jsonl') as f:
    logs = [json.loads(l) for l in f]

steps = [l['step'] for l in logs]
loss_total  = [l['loss/total']   for l in logs]
loss_sdf    = [l['loss/sdf']     for l in logs]
loss_normal = [l['loss/normal']  for l in logs]
loss_eik    = [l['loss/eikonal'] for l in logs]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, vals, title in zip(axes.flat,
    [loss_total, loss_sdf, loss_normal, loss_eik],
    ['Total Loss', 'L_SDF', 'L_normal', 'L_eikonal']):
    ax.plot(steps, vals)
    ax.set_title(title); ax.set_xlabel('Step'); ax.grid(True)
plt.tight_layout()
plt.savefig('../results/loss_curves/training_curves.png', dpi=150)
plt.show()
```
